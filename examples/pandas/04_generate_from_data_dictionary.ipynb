```python
from pathlib import Path
from great_generator import generate_from_data_dictionary, load_data_dictionary

Path("dictionary_demo").mkdir(exist_ok=True)
Path("dictionary_demo/customer_dictionary.csv").write_text('''column_name,data_type,nullable,description,allowed_values,min,max,unique
customer_id,integer,false,Unique customer identifier,,,,true
email,string,false,Customer email address,,,,true
status,string,false,Customer account status,ACTIVE|INACTIVE|PENDING,,,
balance,decimal,true,Current balance,,0,10000,
''')

schema = load_data_dictionary("dictionary_demo/customer_dictionary.csv")
df = generate_from_data_dictionary("dictionary_demo/customer_dictionary.csv", rows=500, seed=42)
df.head()
```
