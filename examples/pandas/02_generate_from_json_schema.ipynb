# Generate from JSON Schema

Create synthetic data from an API-style JSON Schema contract.

```python
%pip install great-generator
```

```python
from great_generator import generate_from_json_schema

json_schema = {
    "type": "object",
    "required": ["customer_id", "email"],
    "properties": {
        "customer_id": {"type": "integer"},
        "email": {"type": "string", "format": "email"},
        "signup_date": {"type": "string", "format": "date"},
        "status": {"type": "string", "enum": ["ACTIVE", "INACTIVE", "PENDING"]},
        "balance": {"type": "number", "minimum": 0, "maximum": 10000},
    },
}

df = generate_from_json_schema(json_schema, rows=1000, seed=42)
df.head()
```

```python
assert set(df["status"]).issubset({"ACTIVE", "INACTIVE", "PENDING"})
df.describe(include="all")
```

JSON Schema support covers a practical documented subset. Normalize complex `$ref`, recursive schemas, and advanced combinators before generation.
