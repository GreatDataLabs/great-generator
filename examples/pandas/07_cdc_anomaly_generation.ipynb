```python
from great_generator import generate_cdc, generate_domain

cdc = generate_cdc("banking", table="customers", rows=1000, operations=["insert", "update", "delete"], late_arrival_rate=0.02, duplicate_rate=0.005, seed=42)
cdc.head()
```

```python
dirty = generate_domain("ecommerce", scale="tiny", anomalies={"null_rate": 0.02, "duplicate_rate": 0.01, "late_arrival_rate": 0.02, "outlier_rate": 0.005}, seed=42)
dirty["orders"].head()
```

Anomalies are opt-in. Default generated datasets are designed to be relationally valid unless anomaly settings request otherwise.
