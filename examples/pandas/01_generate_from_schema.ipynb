# Generate from schema with Pandas

Create a local Pandas DataFrame from a compact schema string for demos, QA tests, or notebooks.

```python
%pip install great-generator
```

```python
from great_generator import generate_from_schema, export_dataset

schema = '''
customer_id int,
customer_name string,
email string,
signup_date date,
account_status string,
balance decimal(12,2)
'''

df = generate_from_schema(schema=schema, rows=1000, seed=42)
df.head()
```

```python
from pathlib import Path

Path("output").mkdir(exist_ok=True)
df.to_csv("output/customers.csv", index=False)
df.to_parquet("output/customers.parquet", index=False)

# Optional helper with overwrite protection.
export_dataset(df, "output/customers_helper", format="parquet", overwrite=True)
```

Delete the local `output/` folder when finished. This notebook does not manage credentials or cloud resources.
