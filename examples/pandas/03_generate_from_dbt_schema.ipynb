# Generate from dbt schema.yml

Analytics engineers can generate synthetic data from dbt model metadata.

```python
from pathlib import Path
from great_generator import generate_from_dbt_schema

Path("dbt_demo/models").mkdir(parents=True, exist_ok=True)
Path("dbt_demo/models/schema.yml").write_text('''
version: 2
models:
  - name: customers
    description: Customer dimension table
    columns:
      - name: customer_id
        data_type: integer
        tests: [unique, not_null]
      - name: email
        data_type: string
        tests: [unique, not_null]
      - name: status
        data_type: string
        tests:
          - accepted_values:
              values: [ACTIVE, INACTIVE, PENDING]
      - name: signup_date
        data_type: date
''')

df = generate_from_dbt_schema("dbt_demo/models/schema.yml", model_name="customers", rows=500, seed=42)
df.head()
```
