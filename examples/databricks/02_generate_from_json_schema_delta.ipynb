Generate synthetic data from a JSON Schema contract and write Delta.

```python
from great_generator import generate_from_json_schema

json_schema = {
    "type": "object",
    "properties": {
        "customer_id": {"type": "integer"},
        "email": {"type": "string", "format": "email"},
        "signup_date": {"type": "string", "format": "date"},
        "status": {"type": "string", "enum": ["ACTIVE", "INACTIVE", "PENDING"]},
    },
    "required": ["customer_id", "email"],
}

pdf = generate_from_json_schema(json_schema, rows=100000, seed=42)
sdf = spark.createDataFrame(pdf)
sdf.write.format("delta").mode("overwrite").saveAsTable("default.synthetic_json_customers")
```
