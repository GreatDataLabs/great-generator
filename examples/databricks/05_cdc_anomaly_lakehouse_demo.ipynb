```python
from great_generator import generate_cdc, generate_domain

cdc_pdf = generate_cdc("banking", table="customers", rows=50000, late_arrival_rate=0.02, duplicate_rate=0.005, seed=42)
dirty = generate_domain("ecommerce", scale="small", anomalies={"null_rate": 0.02, "duplicate_rate": 0.01}, seed=42)

spark.createDataFrame(cdc_pdf).write.format("delta").mode("overwrite").save("dbfs:/tmp/great-generator/bronze/cdc_customers")
spark.createDataFrame(dirty["orders"]).write.format("delta").mode("overwrite").save("dbfs:/tmp/great-generator/bronze/ecommerce_orders_dirty")
```
