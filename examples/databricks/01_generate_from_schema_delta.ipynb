# Databricks: generate from schema and write Delta

Create synthetic customer data and write it as a Delta table.

```python
# Databricks notebook source
%pip install great-generator
```

```python
from great_generator import generate_from_schema

schema = '''
customer_id int,
customer_name string,
email string,
signup_date date,
account_status string,
balance decimal(12,2)
'''

pdf = generate_from_schema(schema=schema, rows=100000, seed=42)
sdf = spark.createDataFrame(pdf)
sdf.show(5)
```

```python
sdf.write.format("delta").mode("overwrite").saveAsTable("default.synthetic_customers")

# For Unity Catalog, use a catalog.schema.table name when configured:
# sdf.write.format("delta").mode("overwrite").saveAsTable("catalog.schema.synthetic_customers")
```
