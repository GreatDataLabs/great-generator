# Generate Parquet for Snowflake

Create synthetic Parquet files that can be loaded through a Snowflake stage.

```python
%pip install great-generator
```

```python
from pathlib import Path
from great_generator import generate_from_schema

schema = '''
customer_id int,
customer_name string,
email string,
signup_date date,
account_status string,
balance decimal(12,2)
'''

df = generate_from_schema(schema, rows=100000, seed=42)
Path("output/snowflake/customers").mkdir(parents=True, exist_ok=True)
df.to_parquet("output/snowflake/customers/customers.parquet", index=False)
df.head()
```
