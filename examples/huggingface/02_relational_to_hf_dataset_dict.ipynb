# Relational output to Hugging Face DatasetDict

Generate related tables and convert each table into a named entry in a `DatasetDict`.

```python
%pip install "great-generator[hf]"
```

```python
from great_generator import generate_relational, to_hf_dataset_dict

tables = {
    "customers": "customer_id int, customer_name string, email string",
    "orders": "order_id int, customer_id int, order_date date, order_total double",
}
relationships = ["customers.customer_id -> orders.customer_id"]
data = generate_relational(tables=tables, relationships=relationships, rows={"customers": 100, "orders": 500}, seed=42)
dataset_dict = to_hf_dataset_dict(data)
dataset_dict
```
