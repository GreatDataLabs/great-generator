# Transformers advisor schema plan

Use a local or cached Transformers model as a design-time advisor. The model proposes a plan; Great Generator generates deterministic rows from the saved plan and seed. The model does not generate row data.

```python
%pip install "great-generator[transformers]"
```

```python
from great_generator import generate_from_schema, infer_generation_plan

schema = """
customer_id int,
customer_name string,
email string,
signup_date date,
account_status string
"""

# By default, the Transformers advisor loads local files only.
# Pre-download/cache the model or use a local model path before running this cell.
plan = infer_generation_plan(schema, advisor="transformers:google/flan-t5-small")
plan.to_dict()
```

```python
df = generate_from_schema(schema, rows=1000, plan=plan, seed=42)
df.head()
```
