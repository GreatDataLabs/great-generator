```python
from great_generator import generate_relational

data = generate_relational(
    tables={
        "dim_customer": {"schema": "customer_id int primary key, region string, customer_segment string", "rows": 5000},
        "dim_product": {"schema": "product_id int primary key, product_category string", "rows": 500},
        "fact_sales": {"schema": "sales_id int primary key, customer_id int references dim_customer.customer_id, product_id int references dim_product.product_id, business_date date, sales_amount double", "rows": 100000},
    },
    relationships=["fact_sales.customer_id -> dim_customer.customer_id", "fact_sales.product_id -> dim_product.product_id"],
    seed=42,
)

for table_name, pdf in data.items():
    spark.createDataFrame(pdf).write.mode("overwrite").format("delta").saveAsTable(table_name)
```
